```python
X = torch.tensor([
    [25, 50000, 650],
    [40, 80000, 720]
], dtype=torch.float32)
X.shape
```

```python
from torch.utils.data import TensorDataset
```

```python
y = torch.tensor([0, 1])
```

```python
dataset = TensorDataset(X, y)
```

```python
len(dataset)
```

```python
dataset[0]
```

```python
from torch.utils.data import DataLoader
loader = DataLoader(dataset, batch_size=1)
```

```python
for X_batch, y_batch in loader:
    print(X_batch)
    print(y_batch)
```
