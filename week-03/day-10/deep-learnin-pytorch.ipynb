```python
class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(3, 2)
        self.relu = nn.ReLU()
        self.layer2 = nn.Linear(2, 1)

    def forward(self, x):
         x = self.layer1(x)
         x = self.relu(x)
         x = self.layer2(x)
         return x

model = NeuralNetwork()
output = model(X)
output  
```

```python
print(output.shape)
y = y.reshape(-1, 1)
print(y.shape)
```

```python
loss_fn = nn.MSELoss()
loss = loss_fn(output, y)
loss
```

### Feature scaling sadece klasik ML'de değil, neural network eğitiminde de önemlidir.
X.mean(dim=0) → her feature'ın ortalaması , 
X.std(dim=0) → her feature'ın standart sapması , 
sonuç → feature'ların ölçekleri birbirine yaklaştırılmış olur.

```python
X_scaled = (X - X.mean(dim=0)) / X.std(dim=0)
X_scaled
```

```python
model = NeuralNetwork()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

for epoch in range(100):
    optimizer.zero_grad()
    output = model(X_scaled)
    loss = loss_fn(output, y)
    loss.backward()
    optimizer.step()

    if epoch % 10 == 0:
        print(epoch, loss.item())
```
